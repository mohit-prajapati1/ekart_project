```python
from pyspark.sql.functions import explode
```

```python
path = 'abfss://source-adls@adlsmewalth.dfs.core.windows.net/ecommerce_financials_initial.json'
df = spark.read.format('json').option("Multiline","True").load(path)
```

```python
metadata_df = df.select("metadata.generated_at","metadata.label","metadata.total_records")
```

```python
profit_loss_df = df.select(explode("profit_loss").alias("PL")).select(
    "PL.cost_price",
    "PL.gst_rate",
    "PL.net_profit",
    "PL.order_id",
    "PL.payout_status",
    "PL.platform_fee",
    "PL.product_id",
    "PL.selling_price",
    "PL.tax",
    "PL.vendor_id",
    "PL.commission_amount"
)
profit_loss_df.write.format("delta").apend
commission_df = df.select(explode("commissions").alias("commission")).select(
    "commission.commission_amount",
    "commission.commission_percentage",
    "commission.order_date",
    "commission.order_id",
    "commission.product_id",
    "commission.vendor_id"
)

vendor_payouts_df = df.select(explode("vendor_payouts").alias("VP")).select("VP.*").show()

```
